```python
import numpy as np
```

#### __NORMALIZATION PROBLEM__: 

```python
x = np.random.random((5,5))
print (x)
```

```python
a = x.mean()
```

```python
b = x.std()
```

```python
normal = (x-a)/b
print (normal)
```

```python
np.save ('X_normalized.npy', normal)
```

#### __DIVISIBLE BY 3 PROBLEM__:

```python
k = np.arange(1,101)
i = np.square(k)
g = i.reshape(10,10)

print (g)
```

```python
t = g [g%3==0]
print (t)
```

```python
np.save ('div_by_3.npy', t)
```
